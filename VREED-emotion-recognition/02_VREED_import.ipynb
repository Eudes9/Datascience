```python
# Charger le fichier de labels (valence + arousal)
labels_path = BASE / '03 Self-Reported Questionnaires'

# Lister les fichiers xlsx disponibles
xlsx_files = list(labels_path.glob('*.xlsx'))
print('Fichiers xlsx disponibles :')
for f in xlsx_files:
    print(f'  - {f.name}')

print()
```

```python
# Charger le Post-Exposure Ratings (labels principaux)
# Adapter le nom exact du fichier si nécessaire
post_exposure_file = [f for f in xlsx_files if 'Post' in f.name or 'post' in f.name]

if post_exposure_file:
    df_labels = pd.read_excel(post_exposure_file[0])
    print(f'✓ Labels chargés : {post_exposure_file[0].name}')
    print(f'  Dimensions : {df_labels.shape}')
    print(f'  Colonnes   : {list(df_labels.columns)}')
    display(df_labels.head(5))
else:
    print('Fichier Post-Exposure non trouvé automatiquement.')
    print('Fichiers disponibles :', [f.name for f in xlsx_files])
    # Charger manuellement en décommentant la ligne suivante :
    # df_labels = pd.read_excel(labels_path / 'NOM_DU_FICHIER.xlsx')
```

## 2. Chargement des features Eye Tracking

```python
eye_path = BASE / '04 Eye Tracking Data'

csv_eye = list(eye_path.glob('*.csv'))
print('Fichiers CSV eye tracking :')
for f in csv_eye:
    print(f'  - {f.name}')
print()

if csv_eye:
    # Prendre le fichier features (pas les signaux bruts)
    features_eye_file = [f for f in csv_eye if 'feature' in f.name.lower() or 'Feature' in f.name]
    target = features_eye_file[0] if features_eye_file else csv_eye[0]
    
    df_eye = pd.read_csv(target)
    print(f'✓ Eye Tracking chargé : {target.name}')
    print(f'  Dimensions : {df_eye.shape}')
    print(f'  Colonnes   : {list(df_eye.columns)}')
    display(df_eye.head(5))
else:
    print('Aucun CSV trouvé dans Eye Tracking Data.')
```

## 3. Chargement des features ECG

```python
ecg_path = BASE / '05 ECG-GSR Data'

csv_ecg_gsr = list(ecg_path.glob('*.csv'))
print('Fichiers CSV ECG-GSR :')
for f in csv_ecg_gsr:
    print(f'  - {f.name}')
print()

# Séparer ECG et GSR
ecg_file = [f for f in csv_ecg_gsr if 'ECG' in f.name or 'ecg' in f.name.lower()]
gsr_file = [f for f in csv_ecg_gsr if 'GSR' in f.name or 'gsr' in f.name.lower()]

if ecg_file:
    df_ecg = pd.read_csv(ecg_file[0])
    print(f'✓ ECG chargé : {ecg_file[0].name}')
    print(f'  Dimensions : {df_ecg.shape}')
    print(f'  Colonnes   : {list(df_ecg.columns)}')
    display(df_ecg.head(5))
else:
    print('Fichier ECG non détecté automatiquement.')
```

## 4. Chargement des features GSR

```python
if gsr_file:
    df_gsr = pd.read_csv(gsr_file[0])
    print(f'✓ GSR chargé : {gsr_file[0].name}')
    print(f'  Dimensions : {df_gsr.shape}')
    print(f'  Colonnes   : {list(df_gsr.columns)}')
    display(df_gsr.head(5))
else:
    print('Fichier GSR non détecté automatiquement.')
```
